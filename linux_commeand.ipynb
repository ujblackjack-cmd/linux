#chown
* chown(change owner)은 리눅스/유닉스 시스템에서 파일이나 디렉터리의 소유자(User) 및 소유 그룹(Group)을 변경하는 명령어입니다. 주로 시스템 관리자가 파일 권한을 관리할 때 사용하며, 루트 사용자(root)만 파일의 소유자를 변경한다.

* 기본 구문: chown [옵션] [소유자]:[소유그룹] [파일/디렉터리]
* 소유자만 변경: chown 사용자명 파일명
* 소유자와 그룹 동시 변경: chown 사용자명:그룹명 파일명
* 하위 파일까지 변경 (-R): chown -R 사용자명:그룹명 디렉터리명 (디렉터리 하위 모든 파일/폴더의 소유권 변경)

#chmod
* chmod(change mode)는 리눅스/유닉스 환경에서 파일이나 디렉토리의 접근 권한(퍼미션)을 변경하는 명령어입니다. 소유자(user), 그룹(group), 기타 사용자(others)에 대해 읽기(r, 4), 쓰기(w, 2), 실행(x, 1) 권한을 기호(u+x)나 숫자(755) 조합으로 부여/제한한다

#핵심 사용법
* 심볼릭 모드 (문자): chmod [대상][연산자][권한] 파일명 (예: chmod u+x file.sh -> 소유자에게 실행 권한 추가)
* 절대 모드 (숫자): chmod [3자리숫자] 파일명 (예: chmod 755 file.sh -> 소유자7/그룹5/기타5 권한 설정)

1. 주요 대상 및 권한 기호
* 대상: u (user, 소유자), g (group, 그룹), o (others, 기타 사용자), a (all, 전체)
* 연산자: + (권한 추가), - (권한 제거), = (권한 설정)
* 권한: r (read, 읽기), w (write, 쓰기), x (execute, 실행)
2. 8진수(숫자) 모드 설정
* 권한 합계: r=4, w=2, x=1
* 예시:
chmod 777 file: 모든 사용자에게 모든 권한 부여 (rwxrwxrwx)
chmod 755 file: 소유자는 모든 권한, 그룹/기타는 읽기+실행 권한 (rwxr-xr-x)
chmod 644 file: 소유자는 읽기+쓰기, 그룹/기타는 읽기 전용 (rw-r--r--)

3. 자주 사용하는 명령어 예시
* chmod +x script.sh: 모든 사용자에게 실행 권한 추가
* chmod -R 755 directory: 하위 디렉토리 및 파일까지 포함하여 권한 변경

#ln
* 리눅스 ln (link) 명령어는 파일이나 디렉터리에 대한 링크(바로가기)를 생성하여, 하나의 파일을 여러 이름으로 접근하게 하는 도구입니다. 기본적으로 하드 링크를 생성하며, -s 옵션을 사용하여 심볼릭 링크(바로가기)를 생성

* 기본 문법: ln [옵션] [원본파일] [링크파일]

심볼릭 링크 (Symbolic Link, -s):
* 윈도우의 '바로가기'와 유사
* 원본 파일이 삭제되면 링크 파일은 동작하지 않는다.
* 파일뿐만 아니라 디렉터리 링크도 가능하다.
예: ln -s [원본] [링크]

하드 링크 (Hard Link, 기본):
* 원본 파일과 동일한 inode를 공유하는 또 다른 파일 이름을 생성
* 원본을 삭제해도 링크 파일은 남아서 작동
* 파일 내용이 서로 동기화
* 디렉터리 링크는 불가능
예: ln [원본] [링크]

#apt
apt는 우분투, 데비안 등 리눅스에서 패키지를 설치, 업데이트, 삭제하는 강력한 명령줄 도구입니다. 주로 sudo 권한이 필요하며, 주요 명령어로 update(목록 갱신), upgrade(패키지 업그레이드), install(설치), remove(삭제)를 사용하여 의존성 문제를 자동으로 해결
* sudo apt update: 패키지 저장소의 최신 정보(목록)를 업데이트.
* sudo apt upgrade: 설치된 모든 패키지를 최신 버전으로 업그레이드.
* sudo apt install [패키지명]: 지정된 패키지 설치.
* sudo apt install -y [패키지명]: 설치 여부를 묻는 과정 없이 자동으로 설치.
* sudo apt remove [패키지명]: 패키지 삭제(설정 파일 남김).

#find
리눅스 find 명령어는 파일 시스템에서 조건(이름, 크기, 시간, 권한 등)에 맞는 파일이나 디렉터리를 검색하는 강력한 도구이다. 기본 구조는 find [경로] [옵션] [표현식]이며, -name (이름), -type (타입: f-파일, d-디렉터리), -size (크기) 등이 가장 많이 사용된다

* 기본 검색: find . -name "test.txt" (현재 디렉터리에서 test.txt 찾기)
* 파일/디렉터리 구분: find /home -type d -name "data" (/home에서 data라는 디렉터리 검색)
* 크기 기준: find / -size +100M (100MB 이상인 파일 검색)
* 시간 기준: find . -mtime -7 (최근 7일 이내에 수정된 파일)
* 검색 후 명령 실행: find . -name "*.log" -delete (.log 파일 찾아 삭제)
* 권한 검색: find . -perm 777 (모든 권한이 부여된 파일 찾기

#systemctl

systemctl은 리눅스에서 systemd 기반의 서비스와 데몬을 관리하는 핵심 명령어로, 서비스의 시작, 중지, 상태 확인, 활성화/비활성화 등을 수행합니다. systemctl [명령어] [서비스명] 형식을 사용하며, 주로 start, stop, restart, status, enable, disable 명령어를 활용하여 시스템을 제어

 [서비스 상태 확인]:
 * systemctl status [서비스명] (현재 동작 상태, 로그 일부 확인)

[서비스 시작/중지]:
* systemctl start [서비스명] (서비스 즉시 시작)
* systemctl stop [서비스명] (서비스 즉시 중지)
* systemctl restart [서비스명] (서비스 재시작)

[서비스 활성화/비활성화 (부팅 시 자동실행)]:
* systemctl enable [서비스명] (부팅 시 자동 시작 설정)
* systemctl disable [서비스명] (부팅 시 자동 시작 해제)
* systemctl enable --now [서비스명] (부팅 시 자동 시작 설정 및 즉시 시작)

[목록 및 기타]:
* systemctl list-units --type=service (작동 중인 서비스 목록)
* systemctl list-unit-files --type=service (모든 서비스의 활성화 상태)
* systemctl daemon-reload (서비스 파일 수정 후 설정 갱신
